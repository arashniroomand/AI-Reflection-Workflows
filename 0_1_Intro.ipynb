```python
# Import required libraries
from dotenv import load_dotenv
from openai import OpenAI
import os

# Load environment variables from .env file
load_dotenv()

# Optional: Verify environment variables are loaded (for debugging)
print("OPENAI_API_KEY loaded:", os.getenv("OPENAI_API_KEY") is not None)
print("OPENAI_BASE_URL loaded:", os.getenv("OPENAI_BASE_URL") is not None)
```

```python
# Initialize OpenAI client with environment variables
# The client automatically reads OPENAI_API_KEY from environment
# Optional: You can also specify custom base URL if needed
client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),  # Get API key from .env
    base_url=os.getenv("OPENAI_BASE_URL")  # Get base URL from .env (optional)
)

# Verify client is ready
print("✅ OpenAI client initialized successfully")
```

```python
# Create a chat completion request
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system", 
            "content": "You are the best friend of Arash. you are so close. Your name is Mr. BV"
        },
        {
            "role": "user", 
            "content": "Hello mr BV im Arash"
        }
    ]
)

# Extract and display the response
print("Model Response:")
print(response.choices[0].message.content)
```
